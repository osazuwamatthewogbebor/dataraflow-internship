````python
# 3. Program to check whether a number entered by the user is even or odd

number = float(input("Enter a number: "))
print(f"The number you entered is {'even' if number % 2 == 0 else 'odd'}.")
````

````python
# 4. Program that prints the multiplication table of a number (from 1–12)

try:
    num = int(input("Enter a number to print its multiplication table: "))
    print(f"Multiplication table for {num}\n", end="\n")
    for i in range(1, 13):
        print(f"{num} x {i:>2} = {num * i:>4}")
except ValueError:
    print("Please enter a valid number.")
````

````python
# 5. Program that calculates the factorial of a number using a loop

try: 
    number = int(input("Enter a number to calculate its factorial: "))

    factorial = number
    for i in range(number - 1, 0, -1):
        factorial *= i
    print(factorial)

except ValueError:
    print("Please enter a valid integer.")
````

````python
# 6. Program that finds the largest number in a given list.
numbers = [1, 4, 33, 56, 2, 46, 7, 2]

max = numbers[0]

for i in numbers:
    if i > max:
        max = i

print(f"The maximum number in this list {numbers} is {max}.")
````

````python
# 7. Program that counts how many vowels are in a given string

# List of vowels
vowels = ["a", "e", "i", "o", "u"]

givenString = "Hello, wElcome to datarAflow!"
count = 0

for char in givenString:
    if char.lower() in vowels:
        count += 1

print(f"There are '{count}' vowels in the given string.")
````

````python
# 8. Function that takes two numbers and returns their sum, difference, and product.

def numberArithmetic(first, second):
    """Takes two numbers as arguments
        Returns their sum, difference and product as a dictionary
    """
    try:
        first = float(first)
        second = float(second)


        # Return a dictionary of sum, difference and product
        return {
            "sum": first + second,
            "difference": first - second,
            "product": round(first * second, 2)
        }
    except ValueError:
        print("Both arguments must be valid numbers!")


# Running the function
numberArithmetic(23, 3)
````

````python
# 9. Program that reverses a string without using slicing (`[::-1]`)

oldString = "Datara Flow!"

newString = "".join(reversed(oldString))
print(newString)
````

````python
# 10. Program that generates the first 10 numbers of the Fibonacci sequence.

fibonacci_seq = [0, 1]

first = 0
next = 1

for i in range(10-2):
    first, next = next, first + next

    fibonacci_seq.append(next)

print("First 10 Fibonacci sequence:",",".join(map(str, fibonacci_seq)))
````

````python
# 1. Student Grades System

print("Enter your Grades (English, Maths, Science, Art, Music)\n")

# Get scores for each subject
try:
  english_grd = float(input("English Grade: "))
  math_grd = float(input("Mathematics Grade: "))
  science_grd = float(input("Science Grade: "))
  art_grd = float(input("Art Grade: "))
  music_grd = float(input("Music Grade: "))


  # Calculate average score
  gradeSum = english_grd + math_grd + science_grd + art_grd + music_grd
  average_score = gradeSum/5
  print("\nAverage Score:", round(average_score, 2))

  # Get the grade
  if average_score < 40:
      print("Grade: ", "F")
  elif average_score < 60:
      print("Grade: ", "C")
  elif average_score < 80:
      print("Grade: ", "B")
  elif average_score <= 100:
      print("Grade: ", "A")
  else:
      print("Invalid Score")
      
except ValueError:
  print("Invalid score")
````

````python
# 2. Calculator program that can perform addition, subtraction, multiplication, and division based on user input

try:
    first_number = float(input("Enter the first number: "))
    operator = input("Enter the operator (+, -, *, /): ")
    second_number = float(input("Enter the second number: "))


    operationDict = {
        "+": lambda a, b: a + b,
        "-": lambda a, b: a - b,
        "*": lambda a, b: a * b,
        "/": lambda a, b: round(a/b, 2) if b != 0 else "Zero division is not allowed!"
    }

    try:
        result = operationDict[operator](first_number, second_number)
        print(first_number, operator, second_number, "=", result)
    except KeyError:
        print("Invalid operator selected. Choose from +, -, *, and /")

except ValueError:
    print("Enter a valid number")
    
````

````python
# 3. Shopping Cart
"""
Program where a user can add items to a shopping cart 
(dictionary with item names and prices), then display the total cost.
"""

shopping_cart = {}

while True:
    item = input("Enter an item (or 'done' to finish): ")

    if item.lower() == "done":
        break

    try:
        price = float(input(f"Enter the price of {item}: ₦"))
        shopping_cart[item] = price

    except ValueError:
        print("Invalid price. Please enter a number.")


print("\nShopping Cart\n")

total = 0

for item, price in shopping_cart.items():
    print(f"{item}: ₦{price:.2f}")
    total += price

print(f"\nTotal cost: ₦{total:.2f}")
````

````python
# Guessing Game

import random

# Generate computer guess
computer = random.randint(1, 20)

while True:
    try:
        # Get user guess
        user = int(input("Enter an integer between 1 - 20 (or 0 to end the game): "))

        # exit code
        if user == 0:
            print("Goodbye!")
            break

        elif user == computer:
            print("Awesome, you guessed it!")
            break

        # Hints
        elif user < computer:
            print("Too low. Aim higher!")
        else:
            print("Too high. Aim lower!")

    except ValueError:
        print("Enter a valid integer!")
````

````python
# File Handling

def main():
    """Program that accepts a list of student names from the user, 
        saves them in a text file, and then reads and displays them.
    """

    print("--- Student Name Manager ---")
    print("Enter student names one by one. Type 'done' when you are finished.")

    students = []

    # Get student names
    while True:
        student_name = input("Enter a student name (or 'done' if you are done): ").strip()

        # Stop condition
        if student_name.lower() == "done":
            break

        if student_name:
            students.append(student_name)
        else:
            print("Name cannot be empty. Please try again.")
            

    if not students:
        print("\nNo names were entered. Program exiting...")
        return


    # Save student names
    filename = "students.txt"
    try:
        with open(filename, "w", encoding="utf-8") as file:
            for student in students:
                file.write(student + "\n")

            print(f"Successfully saved {len(students)} name(s) to '{filename}'")
    except IOError as e:
        print(f"Error writing to file: {e}")
        return

    # Read and display the names back from the file
    print("\n--- Reading and Displaying from File ---")
    try:
        with open(filename, "r", encoding="utf-8") as file:
            saved_students = [line.strip() for line in file]

            print(f"Total students found: {len(saved_students)}")
            for index, student in enumerate(saved_students, start=1):
                print(f"{index}. {student}")

    except FileNotFoundError:
        print(f"Error: Could not find the file '{filename}")

    except IOError as e:
        print(f"Error reading from file: {e}")

if __name__ == "__main__":
    main()
````

#### 4\. What are Python’s data types? Give examples.

 Python data types define the **type of value** stored in a variable.

 a. **Integer:** Integers are whole numbers, such as `10` or `25`.

 b. **Float:** Floats are decimal numbers, such as `3.14` or `2.5`.

 c. **String:** Strings are text values, such as `"Hello"` or `"67"`.

 d. **Boolean:** Boolean values represent `True` or `False`.

 e. **List:** Lists store multiple values and can be changed, such as `[1, 2, 3]`.

 f. **Tuple:** Tuples store multiple values that cannot be changed, such as `(1, 2, 3)`.

 g. **Set:** Sets store unique values and do not allow duplicates, such as `{1, 2, 3}`.

 h. **Dictionary:** Dictionaries store data as key-value pairs, such as `{"name": "John"}`.

 i. **None:** `None` represents the absence of a value, such as `x = None`.

````python
def even_numbers(numbers):
    return list(filter(lambda x : x % 2 == 0, numbers))

print(even_numbers([1, 2, 3, 4, 5, 6, 7, 10]))
````

#### 14\. Write a program to check if a string is a palindrome.

A **palindrome** is a word or string that reads the same forward and backward, such as `"madam"`.

````python
# Palindrome checker

text = input("Enter a string: ").strip()

if text == "":
  print("Text cannot be empty")
elif text == text[::-1]:
    print(f"{text} is a palindrome.")
else:
    print(f"{text} is not a palindrome.")
````
