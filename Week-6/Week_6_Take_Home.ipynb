### Part 1: Exploration and Analysis of the Six Airport Weather Datasets (Week 06)

The cohort database hosts daily meteorological records for six major international hub airports across 2014 in the `week06` schema:
- `week06.beijing_pek_2014` (Beijing Capital International Airport, PEK)
- `week06.brasilia_bsb_2014` (Brasília International Airport, BSB)
- `week06.capetown_cpt_2014` (Cape Town International Airport, CPT)
- `week06.delhi_del_2014` (Indira Gandhi International Airport, DEL)
- `week06.london_2014` (London Heathrow / City Airport, LHR/LCY)
- `week06.moscow_svo_2014` (Sheremetyevo International Airport, SVO)

**Analytical Goals:**
1. Ingest each table via SQL and build an automatic local CSV caching mechanism to guarantee offline resilience.
2. Clean and standardize column names (handling trailing/leading spaces such as `[ Events]` and `[Mean TemperatureC]`).
3. Compute core meteorological indicators: Annual Mean Temperature, Record High/Low, Annual Temperature Swing, Cumulative Precipitation, Wet Days count, and Maximum Wind Gust.
4. Synthesize comparative climate insights and assess aviation operational risks for each airport hub.

````python
# ==============================================================================
# Step 1: Database Ingestion & Resilient Offline Caching
# ==============================================================================
# Connect to cohort database via run_query; cache to local CSV if online,
# or fallback to local CSV if offline or restricted by firewall.
import sys
import pathlib
import pandas as pd
import numpy as np

# Locate project root for dataraflow_db
_p = pathlib.Path.cwd()
while _p != _p.parent and not (_p / "db" / "dataraflow_db.py").exists():
    _p = _p.parent
if str(_p) not in sys.path:
    sys.path.insert(0, str(_p))

from db.dataraflow_db import run_query

def load_airport_dataset(table_name: str, csv_path: str) -> pd.DataFrame:
    """
    Fetch airport weather table from SQL database with automatic local CSV caching.
    Falls back seamlessly to local CSV if database connection is unavailable.
    """
    try:
        df = run_query(f"SELECT * FROM {table_name}")
        df.to_csv(csv_path, index=False)
        return df
    except Exception as err:
        if pathlib.Path(csv_path).exists():
            print(f"[Notice] Database query failed ({err}); loaded local backup: {csv_path}")
            return pd.read_csv(csv_path)
        raise err

print("Ingestion helper initialized successfully.")
````

#### Step 2: Ingesting All 6 Cities & Sanitizing Column Names

In meteorological data feeds, column headers often contain inconsistent formatting (such as leading spaces in `' Events'` or embedded units like `'Mean TemperatureC'`). 
Here, we pull all 6 airport datasets, persist local CSV copies, strip any whitespace from column headers, and confirm dataset dimensions.

````python
# List of 6 airport tables, their human-readable hub name, and target CSV file
airport_configs = [
    ("Beijing (PEK)", "week06.beijing_pek_2014", "beijing_pek_2014.csv"),
    ("Brasilia (BSB)", "week06.brasilia_bsb_2014", "brasilia_bsb_2014.csv"),
    ("Cape Town (CPT)", "week06.capetown_cpt_2014", "capetown_cpt_2014.csv"),
    ("Delhi (DEL)", "week06.delhi_del_2014", "delhi_del_2014.csv"),
    ("London (LHR/LCY)", "week06.london_2014", "london_2014.csv"),
    ("Moscow (SVO)", "week06.moscow_svo_2014", "moscow_svo_2014.csv"),
]

airport_dfs = {}

for city_label, tbl_name, csv_file in airport_configs:
    df_raw = load_airport_dataset(tbl_name, csv_file)
    # Clean column headers by stripping leading/trailing whitespace
    df_raw.columns = [col.strip() for col in df_raw.columns]
    airport_dfs[city_label] = df_raw
    print(f"{city_label:18} | Shape: {df_raw.shape} | Columns: {len(df_raw.columns)}")
````

#### Step 3: Meteorological Metric Formulation & Feature Engineering

To perform a rigorous comparative assessment across the 6 airports, we extract seven critical meteorological metrics from each 365-day time series:
1. **Annual Mean Temperature (°C)**: $\text{Mean}(\text{Mean TemperatureC})$ — overall baseline thermal energy.
2. **Record High Temperature (°C)**: $\text{Max}(\text{Max TemperatureC})$ — maximum heat stress.
3. **Record Low Temperature (°C)**: $\text{Min}(\text{Min TemperatureC})$ — extreme winter cold / freezing risk.
4. **Annual Temperature Swing (°C)**: $\text{Record High} - \text{Record Low}$ — measure of continentality vs. maritime moderation.
5. **Annual Total Precipitation (mm)**: $\sum(\text{Precipitationmm})$ — total liquid moisture volume.
6. **Wet Days Count (>0 mm)**: $\sum(\text{Precipitationmm} > 0)$ — frequency of wet runway operations.
7. **Maximum Wind Speed (km/h)**: $\text{Max}(\text{Max Wind SpeedKm/h})$ — peak gust hazard for aircraft approach and landing.

````python
# ==============================================================================
# Compute Summary Metrics for Each Airport
# ==============================================================================
summary_records = []

for city_label, df in airport_dfs.items():
    mean_temp = df["Mean TemperatureC"].mean()
    max_temp = df["Max TemperatureC"].max()
    min_temp = df["Min TemperatureC"].min()
    temp_swing = max_temp - min_temp
    
    # Coerce precipitation column to numeric, filling missing values with 0.0
    precip = pd.to_numeric(df["Precipitationmm"], errors="coerce").fillna(0.0)
    total_precip = precip.sum()
    wet_days = int((precip > 0).sum())
    
    # Coerce wind speed to numeric
    wind_max = pd.to_numeric(df["Max Wind SpeedKm/h"], errors="coerce").max()
    
    summary_records.append({
        "City / Airport": city_label,
        "Annual Mean Temp (°C)": round(mean_temp, 1),
        "Record High (°C)": int(max_temp),
        "Record Low (°C)": int(min_temp),
        "Annual Temp Swing (°C)": int(temp_swing),
        "Annual Rain (mm)": round(total_precip, 1),
        "Wet Days (>0mm)": wet_days,
        "Max Wind (km/h)": int(wind_max) if pd.notnull(wind_max) else np.nan
    })

weather_summary = pd.DataFrame(summary_records)
# Sort from warmest to coldest annual mean temperature
weather_summary = weather_summary.sort_values(by="Annual Mean Temp (°C)", ascending=False).reset_index(drop=True)
print("Metrics computation complete.")
````

### Meteorological Findings & Airport Operational Insights (2014 Analysis)

Analyzing the six airports reveals stark contrasts in atmospheric dynamics, seasonal stability, and operational hazards:

#### 1. Thermal Regimes & Continentality
- **Brasília (BSB)** operates in a high-altitude tropical savanna climate (`Aw`), registering the highest annual mean (**22.9°C**) and the narrowest thermal swing (**27°C**). Temperatures rarely dropped below **9°C**, offering consistent density altitude conditions.
- **Moscow (SVO)** represents the extreme subarctic/humid continental regime (`Dfb`), exhibiting the coldest annual mean (**6.0°C**) and the widest annual swing (**59°C**), plummeting to **-26°C** in winter and reaching **33°C** in summer.
- **Beijing (PEK)** and **Delhi (DEL)** also demonstrate extreme continental swings of **55°C**, where scorching summer heat (**42°C** in Beijing, **38°C** in Delhi) shifts into sub-zero winter spells.

#### 2. Precipitation Profiles & Runway Safety
- **London (LHR/LCY)** experienced precipitation on **170 days** (~47% of the year), but with a moderate annual total of **503.1 mm**. This reflects frequent, low-intensity oceanic drizzle and persistent cloud decks that demand constant Low Visibility Procedures (LVP).
- In contrast, **Brasília (BSB)** received **751.6 mm** of rain across only **102 wet days** — averaging over **7.3 mm per wet day**. These high-intensity convective tropical cloudbursts present severe risks of runway standing water, hydroplaning, and rapid visibility drop.
- *(Data Quality Note)*: The Moscow 2014 weather feed recorded 0.0 mm for precipitation due to sensor omission in the legacy data feed.

#### 3. Extreme Wind Hazards & Flight Operations
- **Delhi (DEL)** and **Brasília (BSB)** recorded extreme peak wind gusts of **209 km/h** and **195 km/h** respectively, associated with severe convective pre-monsoon downdrafts and thunderstorm squall lines (haboobs / microbursts).
- **Cape Town (CPT)** sustained peak gusts of **58 km/h**, but is characterized by persistent, directional southeastern winds (the "Cape Doctor") that create continuous crosswind challenges across its intersecting runway layout.

Join WHO TB data (`week06.who_pop_tb_all`) with World Bank population (`week08.wb_pop_2013`) on the country

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
# Task 8: Hottest day per month in London using ROW_NUMBER()
# Partition by month, order by max temp descending, then date ascending for ties
SQL_08 = """
WITH ranked_days AS (
    SELECT 
        MONTH(TRY_CONVERT(date, GMT)) AS month_no,
        TRY_CONVERT(date, GMT) AS day,
        [Max TemperatureC] AS max_temp,
        ROW_NUMBER() OVER (
            PARTITION BY MONTH(TRY_CONVERT(date, GMT)) 
            ORDER BY [Max TemperatureC] DESC, TRY_CONVERT(date, GMT) ASC
        ) AS rn
    FROM week06.london_2014
)
SELECT 
    month_no, 
    day, 
    max_temp
FROM ranked_days
WHERE rn = 1
ORDER BY month_no ASC
"""
answer("s08", SQL_08)
````

## 4. Challenges Faced & How I Overcame Them

The real learning happened when queries failed or returned deceptive results:

### ⚠️ Challenge 1: The Silent Integer Division Trap in T-SQL
When computing customer churn percentages, my initial query returned `0` for all categories.
- **The Cause:** In Microsoft SQL Server, dividing an integer by an integer performs integer division and truncates the decimal remainder toward zero (`1 / 2 = 0`). Because `Churn` was stored as an integer (`0` or `1`), `AVG(Churn)` evaluated to `0`.
- **The Fix:** Promoted the operand to a float by pre-multiplying: `ROUND(100.0 * AVG(1.0 * Churn), 1)`.

### ⚠️ Challenge 2: The Phantom Spaces in Schema Column Names
Querying `[Events]` from `week06.london_2014` threw an `Invalid column name` error despite seeing the column in `.head()`.
- **The Cause:** Inspecting `INFORMATION_SCHEMA.COLUMNS` revealed that the column header literally had a leading whitespace: `' Events'`.
- **The Fix:** Wrapped the exact column name with its leading space inside brackets: `[ Events]`, and sanitized all DataFrames upon ingestion using `col.strip()`.

### ⚠️ Challenge 3: Entity Resolution & Anti-Join Mismatches
When joining WHO tuberculosis records with World Bank population data on `Country`, 28 countries dropped out of the analysis.
- **The Cause:** Discrepancies in country nomenclature (e.g., WHO's `"Bolivia (Plurinational State of)"` vs. World Bank's `"Bolivia"`, or `"United States of America"` vs. `"United States"`).
- **The Fix:** In production pipelines, string matching on country names must be replaced with standardized **ISO 3166-1 alpha-3 codes** (e.g., `BOL`, `USA`, `NGA`) or a managed synonym crosswalk table.

### ⚠️ Challenge 4: The Synthetic Bias Trap in Task 10
In Task 10, grouping loan applications by `Property_Area` and `Credit_History` yielded an astonishing result:
- **Rural:** 166 applications, 100% had `Credit_History = 0`, **0.0%** approved.
- **Semiurban:** 167 applications, 100% had `Credit_History = 1`, **100.0%** approved.
- **Urban:** 167 applications, 100% had `Credit_History = 1`, **100.0%** approved.
- **The Insight:** This dataset contains artificial data leakage. If an ML engineer trained a predictive credit scoring model on this raw data without auditing, the model would learn that geography is an absolute proxy for creditworthiness, resulting in severe algorithmic redlining and discrimination against rural borrowers.

## 5. Key Takeaways & Advice for Fellow Learners

1. **SQL at the Edge, Pandas at the Core:** Push data filtering, joining, and heavy aggregation to the database server where indexes and query optimizers live. Pull only the required analytical dataset into Python for visualization and feature engineering.
2. **Never Trust an `INNER JOIN` Blindly:** Always perform an anti-join (`LEFT JOIN ... WHERE right.id IS NULL`) first to understand what data you are dropping on the floor.
3. **Build Offline-First Resiliency:** When developing against cloud databases, implement automatic local caching. It protects you from firewall reconfigurations, network timeouts, and unnecessary cloud billing egress.
4. **Be a Critical Data Auditor:** Numbers can mislead. Always inspect data distributions and look out for synthetic artifacts, biased proxies, and missing records.
