However, pandas does provide a mechanism for you to get and display one or more selected columns as a new dataframe in its own right. To do this you need to use a list.

A list in Python consists of one or more items separated by commas and enclosed within square brackets, for example ['Country'] or ['Country', 'Population(1000s)']. This list is then put within outer square brackets immediately after the dataframe’s name, like this:

```python
df[['TB deaths']].head()
```

```python
df[['Country', 'TB deaths']].head()
```

```python
df[['Country', 'Population (1000s)']].head()
```

#### Comparison Operations

```python
df['Country'] >= 'K'
```

```python
df[df['Country'] >= 'K'].head()
```

```python
df[df['Population (1000s)'] > 80000].head()
```

#### Bitwise Operators

```python
(df['Country'] >= 'Nigeria') & (df['Population (1000s)'] > 80000)
```

```python
df[(df['Country'] >= 'Nigeria') & (df['Population (1000s)'] > 80000)]
```

**If the same columns will be used repeatedly in the program, the code becomes more readable if written as follows**

```python
population = df['Population (1000s)']
deaths = df['TB deaths']

df[(population > 80000) & (deaths > 10000)]
```
