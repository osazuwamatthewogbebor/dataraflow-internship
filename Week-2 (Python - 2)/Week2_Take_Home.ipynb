- Create a base class `Shape` with a method `area()` that raises `NotImplementedError`.
- Create subclasses `Rectangle` and `Circle` that implement `area()` appropriately.
- Instantiate both and print their areas.

````python
import math

class Shape:
    def area(self):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, width, height):
        # Validation
        if not isinstance(width, (int, float)):
            raise TypeError("Width must be a number")

        if width <= 0:
            raise ValueError("Width must be greater than zero")

        if not isinstance(height, (int, float)):
            raise TypeError("Height must be a number")

        if height <= 0:
            raise ValueError("Height must be greater than zero")

        self.width = width
        self.height = height

    def area(self):
        return round(self.width * self.height, 2)


class Circle(Shape):
    def __init__(self, radius):
        # Validation
        if not isinstance(radius, (int, float)):
            raise TypeError("Radius must be a number")

        if radius <= 0:
            raise ValueError("Radius must be greater than zero")

        self.radius = radius

    def area(self):
        return round(math.pi * self.radius ** 2, 2)


newRect = Rectangle(6, 8)
newCircle = Circle(12)

for item in (newRect, newCircle):
    print(item.area())
````

- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance):
        if not isinstance(balance, (int, float)):
            raise TypeError("Invalid opening balance")
        if balance < 0:
            raise ValueError("Opening balance must be positive.")

        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if not isinstance(amount, (int, float)):
            return "Invalid amount"
        if amount <= 0:
            return "Amount must be greater than zero."

        self.balance += amount
        return f'Deposit of {amount} was successful. Balance: {self.balance}'

    def withdrawal(self, amount):
        if not isinstance(amount, (int, float)):
            return "Invalid amount"
        if amount <= 0:
            return "Amount must be greater than zero."

        if amount > self.balance:
            return "Insufficient balance"

        self.balance -= amount
        return f'Withdrawal of {amount} was successful. Balance: {self.balance}'


# Savings subclass
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        if not isinstance(interest_rate, (int, float)):
            raise TypeError("Invalid interest rate")
        if interest_rate <= 0:
            raise ValueError("Interest rate must be positive.")

        super().__init__(owner, balance)
        # Interest rate saved as decimal but passed as integer or float
        self.rate = interest_rate / 100

    def add_interest(self):
        # Balance * (1 + 0.20)
        self.balance *= (1 + self.rate)
        return f'{self.rate * 100}% interest added successfully. Balance: {self.balance}'

# 1. Open account
newSavingsAccount = SavingsAccount("Osazuwa Matthew", 1000000, 20)
print(f"Initial Balance: {newSavingsAccount.balance}")

# 2. Deposit money
deposit_receipt = newSavingsAccount.deposit(50000)
print(deposit_receipt)

# 3. Add interest
interest_receipt = newSavingsAccount.add_interest()
print(interest_receipt)
````

````python

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"{self.brand}-{self.model}"


class Car(Vehicle):
    def __init__(self, brand, model, doors):
        if not isinstance(doors, int):
            raise TypeError("Doors must be an integer.")
        if doors <= 0:
            raise ValueError("Doors cannot be less than 1.")

        super().__init__(brand, model)
        self.doors = doors

    def display_info(self):
        return f"{self.brand} {self.model} ({self.doors} doors)"


class ElectricCar(Car):
    def __init__(self, brand, model, doors, battery_capacity):

        # Validate the battery capacity
        if isinstance(battery_capacity, (int, float)):
            if battery_capacity <= 0:
                raise ValueError("Battery capacity must be greater than zero.")
            self.battery_capacity = f"{battery_capacity}kWh"

        elif isinstance(battery_capacity, str):
            clean_capacity = battery_capacity.strip().lower()
            if not clean_capacity.endswith("kwh"):
                raise ValueError("String battery capacity must end with 'kWh'.")

            # Extract the numeric part
            numeric_part = clean_capacity.replace("kwh", "").strip()
            try:
                val = float(numeric_part)
                if val <= 0:
                    raise ValueError
            except ValueError:
                raise ValueError("Battery capacity must contain a valid positive number (e.g., '122kWh' or 122).")

            self.battery_capacity = f"{numeric_part}kWh"

        else:
            raise TypeError("Battery capacity must be a number or a string.")

        super().__init__(brand, model, doors)

    def display_info(self):
        return f"{self.brand} {self.model} ({self.doors} doors) - {self.battery_capacity} Battery"



car1 = ElectricCar("Ferrari", "Luce", 4, "122kWh")
car2 = ElectricCar("Tesla", "Model S", 4, 100)

print(car1.display_info())
print(car2.display_info())

````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.number = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.number > 0:
            x = self.number
            self.number -= 1

            return x
        else:
            raise StopIteration

counter = CountDown(9)
counter_iter = iter(counter)

for x in counter_iter:
    print(x)
````

**Why len() works with different data types**

The len() function is a built-in function that demonstrates of polymorphism in Python. It works across completely different data types because of Python's dunder methods.

What happens under the hood is that len(object) automatically searches for and runs a special method called __len__() defined inside that object's class.

Create an iterator class `EvenNumbers` that takes a limit and iterates over even numbers only. Use it in a `for` loop.

````python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.number = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.number <= self.limit:
            if self.number % 2 == 0:
                current_number = self.number
                self.number += 1
                return current_number
            else:
                self.number += 1
        raise StopIteration

even_numbers = EvenNumbers(15)

for even_number in even_numbers:
    print(even_number)
````

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

````python
class Flyer:
    def move(self):
        return "Flying"

class Swimmer:
    def move(self):
        return "Swimming"

class Duck(Swimmer, Flyer):
    def move(self):
      return super().move()

duck = Duck()
print(duck.move())
````

**Method Resolution Order (MRO)**

The Method Resolution Order (MRO) is the deterministic set of rules Python follows to search for methods or attributes in a class hierarchy, especially during multiple inheritance.

When super().move() is called inside Duck, Python inspects Duck's MRO sequence: Duck, Flyer, Swimmer. Left-to-Right

Python resolves base classes in the order they are listed in the class definition (class Duck(Swimmer, Flyer)). And because Swimmer comes before Flyer, Python finds and executes Flyer.move(), returning "Flying".

Python uses the C3 Linearization algorithm to construct this order, ensuring that a child class is always checked before its parents and that parent ordering is strictly preserved without duplicate lookups.

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
  def __init__(self, n):
    self.n = n
    self.a, self.b = 0, 1

  def __iter__(self):
    return self

  def __next__(self):
    if self.a > self.n:
      raise StopIteration

    result = self.a
    self.a, self.b = self.b, self.a + self.b
    return result

fib = Fibonacci(100)

for num in fib:
  print(num)
````

  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
# Books Domain
class Book:

    def __init__(self, book_id, title, author, year):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.year = year
        self._is_borrowed = False

    @property
    def is_borrowed(self):
        """Getter for protected borrow status."""
        return self._is_borrowed

    @is_borrowed.setter
    def is_borrowed(self, status):
        """Setter to enforce boolean assignment."""
        if not isinstance(status, bool):
            raise ValueError("is_borrowed must be a boolean value")
        self._is_borrowed = status

    def get_description(self):
        """Base description method for polymorphism."""
        return f"[ID: {self.book_id}] '{self.title}' by {self.author} ({self.year})"

    def __str__(self):
        return self.get_description()


class EBook(Book):

    def __init__(self, book_id, title, author, year, file_size, format_type):
        super().__init__(book_id, title, author, year)
        self.file_size = file_size
        self.format_type = format_type

    def get_description(self):
        return f"{super().get_description()} [EBook: {self.file_size}MB, {self.format_type}]"


class PrintedBook(Book):

    def __init__(self, book_id, title, author, year, weight, dimensions):
        super().__init__(book_id, title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"{super().get_description()} [Printed - {self.weight}g, {self.dimensions}]"


# Membership Domain
class Member:
    __total_borrowed_books = 0

    def __init__(self, name, member_id, limit=1):
        self._name = name
        self._member_id = member_id
        self._borrowed_books = []
        self._limit = limit
        self._iter_index = 0

    @property
    def name(self):
        return self._name

    @property
    def member_id(self):
        return self._member_id

    @property
    def limit(self):
        return self._limit

    @property
    def borrowed_books(self):
        return list(self._borrowed_books)

    @classmethod
    def get_total_borrowed_books(cls):
        return cls.__total_borrowed_books

    def get_account_status(self):
        remaining = self._limit - len(self._borrowed_books)
        return f"{self.name} (ID: {self.member_id}) | Borrowed: {len(self._borrowed_books)}/{self._limit} | Remaining: {remaining}"

    def borrow_book(self, book_obj):
        if book_obj.is_borrowed:
            print(f"Cannot borrow '{book_obj.title}' (ID: {book_obj.book_id}): Already checked out.")
            return False

        if len(self._borrowed_books) >= self.limit:
            print(f"Cannot borrow '{book_obj.title}': {self.name} has reached the limit ({self.limit}).")
            return False

        book_obj.is_borrowed = True
        self._borrowed_books.append(book_obj)
        Member.__total_borrowed_books += 1
        print(f"Successfully borrowed '{book_obj.title}' (ID: {book_obj.book_id}) for {self.name}.")
        return True

    def return_book(self, book_obj):
        if book_obj in self._borrowed_books:
            book_obj.is_borrowed = False
            self._borrowed_books.remove(book_obj)
            Member.__total_borrowed_books -= 1
            print(f"Successfully returned '{book_obj.title}' (ID: {book_obj.book_id}) by {self.name}.")
            return True
        print(f"Return failed: '{book_obj.title}' not found in {self.name}'s account.")
        return False

    def __iter__(self):
        self._iter_index = 0
        return self

    def __next__(self):
        if self._iter_index < len(self._borrowed_books):
            book = self._borrowed_books[self._iter_index]
            self._iter_index += 1
            return book
        else:
            raise StopIteration

    def __str__(self):
        return f"{self.name} (ID: {self.member_id})"


class StudentMember(Member):

    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id, limit=2)
        self.grade_level = grade_level


class TeacherMember(Member):

    def __init__(self, name, member_id, department):
        super().__init__(name, member_id, limit=5)
        self.department = department


# Library Domain
class Library:

    def __init__(self, name):
        self.__library_name = name
        self.__books = []
        self.__members = []
        self.__iter_index = 0

    @property
    def library_name(self):
        return self.__library_name

    @property
    def books(self):
        return list(self.__books)

    @property
    def members(self):
        return list(self.__members)

    def add_book(self, book_obj):
        if isinstance(book_obj, Book):
            # Check for duplicate book IDs
            if any(b.book_id == book_obj.book_id for b in self.__books):
                print(f"Error: Book ID '{book_obj.book_id}' already exists in library.")
                return False
            self.__books.append(book_obj)
            return True

    def add_member(self, member_obj):
        if isinstance(member_obj, Member):
            # Check for duplicate member IDs
            if any(m.member_id == member_obj.member_id for m in self.__members):
                print(f"Error: Member ID '{member_obj.member_id}' already exists.")
                return False
            self.__members.append(member_obj)
            print(f"Registered member: {member_obj}")
            return True

    def remove_book(self, book_obj):
        if book_obj in self.__books:
            if book_obj.is_borrowed:
                for member in self.__members:
                    if book_obj in member.borrowed_books:
                        member.return_book(book_obj)
                        break
            self.__books.remove(book_obj)
            print(f"Removed book '{book_obj.title}' (ID: {book_obj.book_id}) from system.")

    def remove_member(self, member_obj):
        if member_obj in self.__members:
            if len(member_obj.borrowed_books) > 0:
                print(f"Cannot remove {member_obj.name} (ID: {member_obj.member_id}): Unreturned books present.")
                return False
            self.__members.remove(member_obj)
            print(f"Successfully removed member {member_obj.name}.")
            return True
        return False

    def search_book_by_id(self, book_id):
        for book in self.__books:
            if book.book_id == book_id:
                return book
        return None

    def search_member_by_id(self, member_id):
        for member in self.__members:
            if member.member_id == member_id:
                return member
        return None

    def search_book(self, query):
        q = query.lower()
        return [
            book
            for book in self.__books
            if q in book.title.lower()
            or q in book.author.lower()
            or q in str(book.book_id).lower()
        ]

    def __iter__(self):
        self.__iter_index = 0
        return self

    def __next__(self):
        if self.__iter_index < len(self.__books):
            book = self.__books[self.__iter_index]
            self.__iter_index += 1
            return book
        else:
            raise StopIteration


# Running the system
if __name__ == "__main__":
    cohort_library = Library("DataraFlow Library")

    # Instantiate books with unique IDs
    b1 = EBook("B101", "Python 101", "Jane Doe", 2021, 5, "PDF")
    b2 = PrintedBook("B102", "Algorithms", "John Smith", 2018, 1200, "20x25 cm")
    b3 = EBook("B103", "Fluent Python", "Luciano Ramalho", 2021, 15, "EPUB")

    cohort_library.add_book(b1)
    cohort_library.add_book(b2)
    cohort_library.add_book(b3)

    # Instantiate members with unique IDs
    student = StudentMember("Osazuwa", "S001", "DataScience")
    cohort_library.add_member(student)

    # Perform Borrow Operations
    print("\n--- Borrowing Operations ---")
    student.borrow_book(b1)

    # Display status and ID Lookup
    print("\n--- Member Status & Search by ID ---")
    print(student.get_account_status())

    found_book = cohort_library.search_book_by_id("B101")
    print(f"Searched Book ID 'B101': {found_book}")

    # Search by Title
    print("\n--- Searching Books by Title ---")
    title_matches = cohort_library.search_book("Python")
    print(f"Found {len(title_matches)} book(s) matching title 'Python':")
    for book in title_matches:
        print(f"  - {book}")

    # Search by Author
    print("\n--- Searching Books by Author ---")
    author_matches = cohort_library.search_book("John Smith")
    print(f"Found {len(author_matches)} book(s) matching author 'John Smith':")
    for book in author_matches:
        print(f"  - {book}")
````
