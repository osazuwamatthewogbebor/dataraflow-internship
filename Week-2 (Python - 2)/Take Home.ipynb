- Create a base class `Shape` with a method `area()` that raises `NotImplementedError`.
- Create subclasses `Rectangle` and `Circle` that implement `area()` appropriately.
- Instantiate both and print their areas.

````python
import math

class Shape:
    def area(self):
        raise NotImplementedError

class Rectangle(Shape):
    def __init__(self, width, height):
        # Validation
        if not isinstance(width, (int, float)):
            raise TypeError("Width must be a number")

        if width <= 0:
            raise ValueError("Width must be greater than zero")

        if not isinstance(height, (int, float)):
            raise TypeError("Height must be a number")

        if height <= 0:
            raise ValueError("Height must be greater than zero")
            
        self.width = width
        self.height = height

    def area(self):
        return round(self.width * self.height, 2)


class Circle(Shape):
    def __init__(self, radius):
        # Validation
        if not isinstance(radius, (int, float)):
            raise TypeError("Radius must be a number")

        if radius <= 0:
            raise ValueError("Radius must be greater than zero")

        self.radius = radius

    def area(self):
        return round(math.pi * self.radius ** 2, 2)


newRect = Rectangle(6, 8)
newCircle = Circle(12)

for item in (newRect, newCircle):
    print(item.area())
````

- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance):
        if not isinstance(balance, (int, float)):
            raise TypeError("Invalid opening balance")
        if balance < 0:
            raise ValueError("Opening balance must be positive.")

        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if not isinstance(amount, (int, float)):
            return "Invalid amount"
        if amount <= 0:
            return "Amount must be greater than zero."

        self.balance += amount
        return f'Deposit of {amount} was successful. Balance: {self.balance}'

    def withdrawal(self, amount):
        if not isinstance(amount, (int, float)):
            return "Invalid amount"
        if amount <= 0:
            return "Amount must be greater than zero."

        if amount > self.balance:
            return "Insufficient balance"
            
        self.balance -= amount
        return f'Withdrawal of {amount} was successful. Balance: {self.balance}'


# Savings subclass
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        if not isinstance(interest_rate, (int, float)):
            raise TypeError("Invalid interest rate")
        if interest_rate <= 0:
            raise ValueError("Interest rate must be positive.")
            
        super().__init__(owner, balance)
        # Interest rate saved as decimal but passed as integer or float
        self.rate = interest_rate / 100 

    def add_interest(self):
        # Balance * (1 + 0.20)
        self.balance *= (1 + self.rate)
        return f'{self.rate * 100}% interest added successfully. Balance: {self.balance}'

# 1. Open account
newSavingsAccount = SavingsAccount("Osazuwa Matthew", 1000000, 20)
print(f"Initial Balance: {newSavingsAccount.balance}") 

# 2. Deposit money
deposit_receipt = newSavingsAccount.deposit(50000)
print(deposit_receipt) 

# 3. Add interest
interest_receipt = newSavingsAccount.add_interest()
print(interest_receipt)
````

- Create a base class `Vehicle` with attributes `brand` and `model`, and a method `display_info()`.
- Create a subclass `Car` that adds `doors` and overrides `display_info()`.
- Create another subclass `ElectricCar` (child of `Car`) that adds `battery_capacity` and overrides `display_info()` to include all details.
- Create an `ElectricCar` object and call `display_info()`.

````python

class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"{self.brand}-{self.model}"


class Car(Vehicle):
    def __init__(self, brand, model, doors):
        if not isinstance(doors, int):
            raise TypeError("Doors must be an integer.")
        if doors <= 0:
            raise ValueError("Doors cannot be less than 1.")
            
        super().__init__(brand, model)
        self.doors = doors

    def display_info(self):
        return f"{self.brand} {self.model} ({self.doors} doors)"


class ElectricCar(Car):
    def __init__(self, brand, model, doors, battery_capacity):

        # Validate the battery capacity
        if isinstance(battery_capacity, (int, float)):
            if battery_capacity <= 0:
                raise ValueError("Battery capacity must be greater than zero.")
            self.battery_capacity = f"{battery_capacity}kWh"
            
        elif isinstance(battery_capacity, str):
            clean_capacity = battery_capacity.strip().lower()
            if not clean_capacity.endswith("kwh"):
                raise ValueError("String battery capacity must end with 'kWh'.")
                
            # Extract the numeric part
            numeric_part = clean_capacity.replace("kwh", "").strip()
            try:
                val = float(numeric_part)
                if val <= 0:
                    raise ValueError
            except ValueError:
                raise ValueError("Battery capacity must contain a valid positive number (e.g., '122kWh' or 122).")
                
            self.battery_capacity = f"{numeric_part}kWh" 
            
        else:
            raise TypeError("Battery capacity must be a number or a string.")
        
        super().__init__(brand, model, doors)

    def display_info(self):
        return f"{self.brand} {self.model} ({self.doors} doors) - {self.battery_capacity} Battery"



car1 = ElectricCar("Ferrari", "Luce", 4, "122kWh")
car2 = ElectricCar("Tesla", "Model S", 4, 100)

print(car1.display_info())
print(car2.display_info())

````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.number = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.number > 0:
            x = self.number
            self.number -= 1

            return x
        else:
            raise StopIteration

counter = CountDown(9)
counter_iter = iter(counter)

for x in counter_iter:
    print(x)
````

- Create a class `EvenNumbers` that generates even numbers starting from 0.  
- The iterator should be infinite (no `StopIteration`).  
- Demonstrate usage by printing the first 10 even numbers using a loop and `break`.

````python
class EvenNumbers:
    def __iter__(self):
        self.number = 0
        return self

    def __next__(self):
        x = self.number
        self.number += 2

        return x

my_even_numbers = EvenNumbers()

for x in iter(my_even_numbers):
    if x > 10:
        break
    print(x)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def __iter__(self):
        self.file = open(self.filename, "r")
        return self

    def __next__(self):
        line = self.file.readline()
        if not line:
            self.file.close()
            raise StopIteration

        return line


fileReader = FileLineReader("testing.txt")
iter_file_reader = iter(fileReader)

for line in iter_file_reader:
    print(line)
````

- Print the results and explain why `len()` works with different data types.


````python
# len() with a string
my_string = "Python"
print(f"String length: {len(my_string)}") 

# len() with a list
my_list = ["Car", "Bicycle", "ElectricCar"]
print(f"List length: {len(my_list)}")   

# len() with a dictionary
my_dict = {"brand": "Ferrari", "model": "Luce"}
print(f"Dictionary length: {len(my_dict)}") 
````

**Why len() works with different data types**

The len() function is a built-in function that demonstrates of polymorphism in Python. It works across completely different data types because of Python's dunder methods.

What happens under the hood is that len(object) automatically searches for and runs a special method called __len__() defined inside that object's class.

When one passes a string, it calls str.__len__() (counts characters). For a list, it calls list.__len__() (counts items).
For a dictionary, it calls dict.__len__() (counts key-value pairs).

All built-in iterables implement their own unique versions of the __len__() method, the single len() function simply calls them.

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculates and returns the area of the shape"""
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2


shapes = [
    Rectangle(width=5, height=4),
    Circle(radius=3),
    Rectangle(width=10, height=2),
    Circle(radius=5)
]

for shape in shapes:
    print(f"The area of the {type(shape).__name__} is {shape.area():.2f}")
````

- Inside a function `add_to_total(n)`, use the `global` keyword to modify `total` by adding `n`.  
- Call the function multiple times and print the updated value of `total`.

````python
total = 0

def add_to_total(n):
    global total
    total += n

add_to_total(4)
add_to_total(5)
add_to_total(2)

print(total)
````
