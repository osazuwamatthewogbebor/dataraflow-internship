### Create an Iterator

```python
# To create an object/class as an iterator, you have to implement the methods __iter__() and __next__() to your class
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        x = self.a
        self.a += 1
        return x

myclass = MyNumbers()
myiter = iter(myclass)

print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
```

```python
# StopIteration to prevent the iteration from going on forever
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        if self.a <= 20:
            x = self.a
            self.a += 1
            return x
        else:
            raise StopIteration

myclass = MyNumbers()
myiter = iter(myclass)

for x in myiter:
    print(x)
```

```python
# 2. 

class CountDown:
    def __init__(self, n):
        self.number = n

    def __iter__(self):
        return self
        
    def __next__(self):
        if self.number >= 1:
            x = self.number
            self.number -= 1

            return x
        else:
           raise StopIteration

# Implementation
mycounter = CountDown(11)

for x in iter(mycounter):
    print(x)
```

```python
# 3. 

class EvenNumbers:
    def __iter__(self):
        self.number = 0
        return self

    def __next__(self):
        x = self.number
        self.number += 2

        return x

my_even_numbers = EvenNumbers()

for x in iter(my_even_numbers):
    if x > 10:
        break
    print(x)
```

```python
# 4.

class FileLineReader:
    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def __iter__(self):
        self.file = open(self.filename, "r")
        return self

    def __next__(self):
        line = self.file.readline()
        if not line:
            self.file.close()
            raise StopIteration
        return line

fileHandler = FileLineReader("testing.txt")

for line in iter(fileHandler):
    print(line)
        
```
